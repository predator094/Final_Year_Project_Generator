```python
import gradio as gr

def update_state(input_key, input_value, updated_keys, updated_values):
    # Update the set of modified keys
    updated_keys.add(input_key)
    # Update the values dictionary
    updated_values[input_key] = input_value
    return list(updated_keys), updated_values

def reset():
    return [], {}

with gr.Blocks() as demo:
    gr.Markdown("### Gradio App with State Management")
    
    # State variables
    updated_keys = gr.State(set())  # Store updated keys
    updated_values = gr.State({})   # Store updated values
    
    # Input components
    text_input = gr.Textbox(label="Text Input", interactive=True)
    slider_input = gr.Slider(0, 100, label="Slider Input", interactive=True)
    dropdown_input = gr.Dropdown(["Option 1", "Option 2", "Option 3"], label="Dropdown Input", interactive=True)
    checkbox_input = gr.Checkbox(label="Checkbox Input", interactive=True)
    
    # Output components
    updated_keys_display = gr.Textbox(label="Updated Keys", interactive=False)
    updated_values_display = gr.JSON(label="Updated Values")
    
    # Event listeners
    text_input.change(update_state, [gr.Text("text_input",visible=False), text_input, updated_keys, updated_values], [updated_keys_display, updated_values_display])
    slider_input.change(update_state, [gr.Text("slider_input",visible=False), slider_input, updated_keys, updated_values], [updated_keys_display, updated_values_display])
    dropdown_input.change(update_state, [gr.Text("dropdown_input",visible=False), dropdown_input, updated_keys, updated_values], [updated_keys_display, updated_values_display])
    checkbox_input.change(update_state, [gr.Text("checkbox_input",visible=False), checkbox_input, updated_keys, updated_values], [updated_keys_display, updated_values_display])
    
    # Reset button
    reset_button = gr.Button("Reset State")
    reset_button.click(reset, [], [updated_keys_display, updated_values_display])

# Launch the app
demo.launch()
```
